```python
import matplotlib.pyplot as plt
from scipy.io import loadmat

file_path = r"..\dataset\raw_data\A00\A00001.mat"

data = loadmat(file_path)

print("MAT file contents:")
print(data.keys())

ecg = data['val']

print("ECG shape:", ecg.shape)

plt.figure(figsize=(15, 4))
plt.plot(ecg[0])

plt.title("ECG Signal - A00001")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```
